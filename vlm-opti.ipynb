```python
# ==============================================================================
# VLM SOTA: PubMedBERT (Fine-Tuned) + ResNet Grid + Cross-Attention
# ==============================================================================

# --- COMPATIBILITY FIXES ---
import os
os.environ["TF_USE_LEGACY_KERAS"] = "1"
os.environ["TF_KERAS"] = "1"

import numpy as np
import matplotlib.pyplot as plt
import json
import cv2
import sys
import pickle
import time
import random 
from tqdm import tqdm
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score
from sklearn.utils import class_weight
import seaborn as sns
import pandas as pd

import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout, Input, Concatenate, BatchNormalization
from tensorflow.keras.layers import RandomRotation, RandomZoom, RandomTranslation, Reshape, MultiHeadAttention, LayerNormalization, Add, Flatten, GlobalAveragePooling1D
from tensorflow.keras.models import Model
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.regularizers import l2
import tensorflow.keras.backend as K
from tensorflow.keras.optimizers import Adam
from transformers import AutoTokenizer, TFBertModel

# ---  SEED FIXATION (Reproducibility) ---
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    print(f" Random seeds fixed to {seed}")

set_seed(42)

# ---  CONFIGURATION ---
BASE_PATH = 'YOUR_PATH_TO_DATA'
IMAGE_PATH = os.path.join(BASE_PATH, 'VQA_RAD Image Folder') 
JSON_PATH = os.path.join(BASE_PATH, 'VQA_RAD Dataset Public.json')
RESULTS_PATH = './results' 
if not os.path.exists(RESULTS_PATH): os.makedirs(RESULTS_PATH)

IMG_SIZE = (224, 224)
MAX_Q_LEN = 25          
BATCH_SIZE = 32         
LEARNING_RATE = 2e-5    
DROPOUT_RATE = 0.4      
N_FOLDS = 5
MIN_FREQ = 5            

BERT_MODEL_NAME = "microsoft/BiomedNLP-PubMedBERT-base-uncased-abstract-fulltext" 

# ---  DATA LOADING ---
print("--- 1. Class Analysis and Filtering ---")
try:
    with open(JSON_PATH, 'r') as f: raw_data = json.load(f)
except FileNotFoundError:
    sys.exit(f"ERROR: JSON file not found at {JSON_PATH}")

counts = {}
for item in raw_data:
    ans = str(item['answer']).lower()
    counts[ans] = counts.get(ans, 0) + 1

final_vocab = [ans for ans, count in counts.items() if count >= MIN_FREQ]
final_vocab = sorted(final_vocab)
atoi = {w: i for i, w in enumerate(final_vocab)}
NUM_CLASSES = len(final_vocab)

print(f"Number of classes retained (Freq >= {MIN_FREQ}): {NUM_CLASSES}")

X_images, Y_ids, questions_text = [], [], []
filenames = [] 

print("---  Loading Images and Text ---")
for item in tqdm(raw_data):
    ans = str(item['answer']).lower()
    if ans in atoi:
        path = os.path.join(IMAGE_PATH, item['image_name'])
        img = cv2.imread(path)
        if img is not None:
            img = cv2.resize(img, IMG_SIZE)
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            X_images.append(img)
            Y_ids.append(atoi[ans])
            questions_text.append(str(item['question']).lower())
            filenames.append(item['image_name'])

X_images = preprocess_input(np.array(X_images))
Y_ids = np.array(Y_ids)
Y_one_hot = to_categorical(Y_ids, num_classes=NUM_CLASSES)

# ---  TOKENIZATION ---
print(f"---  Tokenization with {BERT_MODEL_NAME} ---")
tokenizer = AutoTokenizer.from_pretrained(BERT_MODEL_NAME)

print("\n TOKENIZER QUALITY CHECK:")
test_words = ["pneumothorax", "cholecystectomy", "pneumomediastinum"]
for w in test_words:
    tokens = tokenizer.tokenize(w)
    print(f"   Word: '{w}' -> Tokens: {tokens}")
print("-" * 60 + "\n")

def bert_encode(texts, tokenizer, max_len=MAX_Q_LEN):
    input_ids = []
    attention_masks = []
    for text in texts:
        encoded = tokenizer.encode_plus(
            text, add_special_tokens=True, max_length=max_len,
            padding='max_length', truncation=True,
            return_attention_mask=True, return_token_type_ids=False
        )
        input_ids.append(encoded['input_ids'])
        attention_masks.append(encoded['attention_mask'])
    return np.array(input_ids), np.array(attention_masks)

X_input_ids, X_attn_masks = bert_encode(questions_text, tokenizer)

# ---  CLASS WEIGHTS & MODEL ---
class_weights_vals = class_weight.compute_class_weight(
    class_weight='balanced', classes=np.unique(Y_ids), y=Y_ids)
class_weights_vals = np.sqrt(class_weights_vals) 
class_weights_vals = np.clip(class_weights_vals, 0.2, 3.0) 
class_weights_dict = dict(enumerate(class_weights_vals))

def build_attention_vlm():
    #  IMAGE BRANCH (ResNet)
    in_img = Input(shape=(224, 224, 3), name="image_input")
    aug = RandomRotation(0.05)(in_img)
    aug = RandomTranslation(0.05, 0.05)(aug)
    
    base_resnet = ResNet50(weights='imagenet', include_top=False, input_tensor=aug)
    base_resnet.trainable = True
    for layer in base_resnet.layers[:-40]: layer.trainable = False 
    
    x_img_grid = Reshape((-1, 2048))(base_resnet.output) 
    x_img_grid = Dense(512, activation='relu', kernel_regularizer=l2(1e-4))(x_img_grid)

    #  TEXT BRANCH (PubMedBERT Fine-Tuning)
    in_ids = Input(shape=(MAX_Q_LEN,), dtype=tf.int32, name="input_ids")
    in_mask = Input(shape=(MAX_Q_LEN,), dtype=tf.int32, name="attention_mask")
    
    biobert_model = TFBertModel.from_pretrained(BERT_MODEL_NAME, from_pt=True)
    
    # FINE-TUNING: Last 4 Layers
    biobert_model.trainable = True
    try:
        bert_layers = biobert_model.layers[0].encoder.layer
        for layer in bert_layers[:-4]: layer.trainable = False
        for layer in bert_layers[-4:]: layer.trainable = True
        print(" PubMedBERT Fine-Tuning: Activated (Last 4 Layers)")
    except:
        print(" Complex BERT structure: Global training with low LR.")
    
    bert_output = biobert_model(input_ids=in_ids, attention_mask=in_mask)[0]
    
    # [CLS] Token Strategy
    cls_token = bert_output[:, 0, :] 
    x_txt_query = Dense(512, activation='relu', kernel_regularizer=l2(1e-4))(cls_token)
    x_txt_query = Reshape((1, 512))(x_txt_query)

    #  CROSS-ATTENTION
    attention_layer = MultiHeadAttention(num_heads=8, key_dim=64) 
    x_attended_img = attention_layer(query=x_txt_query, value=x_img_grid, key=x_img_grid)
    
    x_context = Add()([x_txt_query, x_attended_img])
    x_context = LayerNormalization()(x_context)
    x_context = Flatten()(x_context)
    
    #  CLASSIFICATION HEAD
    x = Dense(512, activation='relu', kernel_regularizer=l2(1e-4))(x_context)
    x = BatchNormalization()(x)
    x = Dropout(DROPOUT_RATE)(x)
    
    output = Dense(NUM_CLASSES, activation='softmax')(x)
    
    model = Model(inputs=[in_img, in_ids, in_mask], outputs=output)
    optimizer = Adam(learning_rate=LEARNING_RATE)
    model.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
    return model

# ---  CROSS-VALIDATION LOOP ---
kfold = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=42)
histories = []
all_true, all_pred = [], []
inference_times = []
fold_scores = [] # For Variance Analysis

print(f"\n--- Starting Cross-Validation ({N_FOLDS} Folds) ---")

fold = 1
for train_idx, val_idx in kfold.split(X_images, Y_ids):
    print(f"\n[Fold {fold}/{N_FOLDS}] -----------------------------")
    K.clear_session()
    
    X_img_tr, X_img_val = X_images[train_idx], X_images[val_idx]
    X_ids_tr, X_ids_val = X_input_ids[train_idx], X_input_ids[val_idx]
    X_msk_tr, X_msk_val = X_attn_masks[train_idx], X_attn_masks[val_idx]
    Y_tr, Y_val = Y_one_hot[train_idx], Y_one_hot[val_idx]
    
    model = build_attention_vlm() 
    
    es = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1)
    
    h = model.fit(
        x=[X_img_tr, X_ids_tr, X_msk_tr],
        y=Y_tr,
        validation_data=([X_img_val, X_ids_val, X_msk_val], Y_val),
        epochs=35,
        batch_size=BATCH_SIZE,
        class_weight=class_weights_dict, 
        callbacks=[es, reduce_lr],
        verbose=1
    )
    histories.append(h.history)
    
    # COMPLEXITY MEASUREMENT (Time)
    print(" Measuring inference time (with Warm-up)...")
    _ = model.predict([X_img_val[:1], X_ids_val[:1], X_msk_val[:1]], verbose=0) # Warm-up
    
    start_time = time.time()
    preds_probs = model.predict([X_img_val, X_ids_val, X_msk_val], verbose=0)
    end_time = time.time()
    
    avg_time_ms = ((end_time - start_time) / len(X_img_val)) * 1000
    inference_times.append(avg_time_ms)
    
    preds = np.argmax(preds_probs, axis=1)
    trues = np.argmax(Y_val, axis=1)
    
    # Fold Accuracy for Variance Analysis
    fold_acc = accuracy_score(trues, preds)
    fold_scores.append(fold_acc)
    
    all_pred.extend(preds)
    all_true.extend(trues)
    fold += 1

# ---  RESULTS & VISUALIZATIONS ---
all_true = np.array(all_true)
all_pred = np.array(all_pred)
model_params = model.count_params()

#  LEARNING CURVES GENERATION
print("\n Generating Learning Curves...")
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
for i, h in enumerate(histories):
    plt.plot(h['accuracy'], linestyle='--', alpha=0.5, label=f'Train Fold {i+1}')
    plt.plot(h['val_accuracy'], linewidth=2, label=f'Val Fold {i+1}')
plt.title('Accuracy per Epoch')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
for i, h in enumerate(histories):
    plt.plot(h['loss'], linestyle='--', alpha=0.5, label=f'Train Fold {i+1}')
    plt.plot(h['val_loss'], linewidth=2, label=f'Val Fold {i+1}')
plt.title('Loss per Epoch')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig(os.path.join(RESULTS_PATH, 'learning_curves.png'))
plt.show()

# B. VARIANCE ANALYSIS BETWEEN FOLDS
print("\n Analyzing Fold Variance (Stability)...")
mean_acc = np.mean(fold_scores)
std_acc = np.std(fold_scores)
print(f"Mean Accuracy across {N_FOLDS} folds: {mean_acc*100:.2f}%")
print(f"Standard Deviation: {std_acc*100:.2f}%")

if std_acc < 0.05:
    print("High Stability: The model performs consistently across data splits.")
else:
    print("Moderate Instability: Performance varies significantly depending on the data split.")

# C. OPEN-ENDED QUESTIONS CONFUSION MATRIX
print("\n Generating Open-Ended Questions Confusion Matrix...")
# Identify Open/Closed
closed_labels = ['yes', 'no']
closed_ids = [atoi[w] for w in closed_labels if w in atoi]
is_closed = np.isin(all_true, closed_ids)
is_open = ~is_closed

# Filter Data for Open Questions Only
open_true = all_true[is_open]
open_pred = all_pred[is_open]

# Get indices and names of classes actually present in Open Questions
unique_indices = np.unique(np.concatenate([open_true, open_pred]))
unique_names = [final_vocab[i] for i in unique_indices]

# Compute Matrix
cm_open = confusion_matrix(open_true, open_pred, labels=unique_indices, normalize='true')

plt.figure(figsize=(16, 16))
sns.heatmap(cm_open, annot=True, fmt='.0%', cmap='Blues', 
            xticklabels=unique_names, yticklabels=unique_names, cbar=False)
plt.title('Detailed Confusion Matrix: Open-Ended Questions Only', fontsize=18)
plt.ylabel('True Answer')
plt.xlabel('Predicted Answer')
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_PATH, 'confusion_matrix_open.png'))
plt.show()

# --- 8. EXECUTIVE SUMMARY ---
# Calculating Metrics
global_acc = accuracy_score(all_true, all_pred)
macro_f1 = f1_score(all_true, all_pred, average='macro')
mean_time = np.mean(inference_times)
acc_closed = accuracy_score(all_true[is_closed], all_pred[is_closed])
acc_open = accuracy_score(all_true[is_open], all_pred[is_open])

print("\n" + "="*60)
print(" EXECUTIVE SUMMARY REPORT")
print("="*60)
print(f" Model Architecture : PubMedBERT (Fine-Tuned) + ResNet50 + Cross-Attention")
print(f" Validation Strategy: {N_FOLDS}-Fold Stratified Cross-Validation")
print("-" * 60)
print(f" GLOBAL PERFORMANCE")
print(f"   • Overall Accuracy : {global_acc*100:.2f}% (± {std_acc*100:.2f}%)")
print(f"   • Macro F1-Score   : {macro_f1:.4f}")
print("-" * 60)
print(f" MEDICAL REASONING (Open vs Closed)")
print(f"   • Closed Questions (Yes/No) : {acc_closed*100:.2f}%")
print(f"   • Open Questions (Diagnosis): {acc_open*100:.2f}%  <-- KEY METRIC")
print("-" * 60)
print(f" EFFICIENCY & COST")
print(f"   • Model Size       : {model_params/1e6:.2f} Million Parameters")
print(f"   • Inference Time   : {mean_time:.2f} ms/sample")
print("="*60)

# Saving Full Data
results_data = {
    'model': 'VLM_PubMedBERT_FT',
    'accuracy': global_acc,
    'std_dev': std_acc,
    'macro_f1': macro_f1,
    'acc_closed': acc_closed,
    'acc_open': acc_open,
    'inference_time_ms': mean_time,
    'params': model_params,
    'fold_scores': fold_scores,
    'all_true': all_true,
    'all_pred': all_pred,
    'classes': final_vocab
}

pkl_path = os.path.join(RESULTS_PATH, 'vlm_sota_results.pkl')
with open(pkl_path, 'wb') as f:
    pickle.dump(results_data, f)
print(f"\n Full results saved to: {pkl_path}")

# Coverage Analysis 
print("\n Generating Dataset Coverage Graph...")
unique_true, counts_true = np.unique(all_true, return_counts=True)
unique_pred, counts_pred = np.unique(all_pred, return_counts=True)
df_true = pd.DataFrame({'Class': [final_vocab[i] for i in unique_true], 'Count': counts_true, 'Type': 'Reality (True)'})
df_pred = pd.DataFrame({'Class': [final_vocab[i] for i in unique_pred], 'Count': counts_pred, 'Type': 'Prediction (Pred)'})
df_cov = pd.concat([df_true, df_pred])
order = df_true.sort_values('Count', ascending=False)['Class'].tolist()
plt.figure(figsize=(20, 8))
sns.barplot(data=df_cov, x='Class', y='Count', hue='Type', order=order, palette={'Reality (True)': 'gray', 'Prediction (Pred)': 'green'})
plt.xticks(rotation=90)
plt.title("Dataset Coverage: Real vs Predicted Distribution")
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_PATH, 'dataset_coverage_analysis.png'))
plt.show()
```
